```python
# ============================================================
# FINAL PROJECT STEP 2B
# Improved Retrieval + Diagnostic Testing
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# ------------------------------------------------------------
# 1. Chunk the documents
# ------------------------------------------------------------

def chunk_text(text, max_words=55, overlap_words=10):
    words = text.split()
    chunks = []
    start = 0

    while start < len(words):
        end = min(start + max_words, len(words))

        chunks.append(
            " ".join(words[start:end])
        )

        if end == len(words):
            break

        start = max(0, end - overlap_words)

    return chunks


chunk_rows = []

for doc in documents:

    doc_chunks = chunk_text(
        doc['text'],
        max_words=55,
        overlap_words=10
    )

    for i, chunk in enumerate(doc_chunks):

        chunk_rows.append({
            'chunk_id': f"{doc['doc_id']}-CHUNK-{i+1}",
            'doc_id': doc['doc_id'],
            'title': doc['title'],
            'chunk_text': chunk
        })


chunks_df = pd.DataFrame(chunk_rows)

print("Total chunks created:", len(chunks_df))


# ------------------------------------------------------------
# 2. Improved searchable text
# ------------------------------------------------------------

# Give the document title extra weight.
chunks_df['search_text'] = (
    chunks_df['title']
    + " "
    + chunks_df['title']
    + " "
    + chunks_df['chunk_text']
)


# ------------------------------------------------------------
# 3. TF-IDF retrieval
# ------------------------------------------------------------

vectorizer = TfidfVectorizer(
    stop_words='english',
    ngram_range=(1, 2)
)

chunk_matrix = vectorizer.fit_transform(
    chunks_df['search_text']
)


def retrieve_chunks(question, top_k=3):

    query_vector = vectorizer.transform([question])

    scores = cosine_similarity(
        query_vector,
        chunk_matrix
    ).flatten()

    top_indices = scores.argsort()[::-1][:top_k]

    results = chunks_df.iloc[top_indices].copy()

    results['score'] = scores[top_indices]

    return results[
        [
            'chunk_id',
            'doc_id',
            'title',
            'chunk_text',
            'score'
        ]
    ]


# ------------------------------------------------------------
# 4. Run all 10 questions for diagnostic review
# ------------------------------------------------------------

diagnostic_rows = []

for _, row in test_cases.iterrows():

    retrieved = retrieve_chunks(
        row['question'],
        top_k=3
    )

    top1 = retrieved.iloc[0]
    top2 = retrieved.iloc[1]

    diagnostic_rows.append({
        'test_id': row['test_id'],
        'question': row['question'],
        'expected_source': row['expected_source'],
        'top1_source': top1['doc_id'],
        'top1_score': round(float(top1['score']), 4),
        'top2_source': top2['doc_id'],
        'top2_score': round(float(top2['score']), 4),
        'score_margin': round(
            float(top1['score'] - top2['score']),
            4
        )
    })


retrieval_diagnostics = pd.DataFrame(
    diagnostic_rows
)


# Save diagnostics
retrieval_diagnostics.to_csv(
    project_path
    / 'results'
    / 'retrieval_diagnostics.csv',
    index=False
)


print("\nRetrieval diagnostics for 10 final test cases:")

retrieval_diagnostics
```

```python
# ============================================================
# FINAL PROJECT STEP 2C
# Improved Retrieval with Domain Metadata
# ============================================================

document_keywords = {
    'ORTHO-001': (
        'office hours opening closing time Friday Saturday '
        'business hours contact schedule open closed'
    ),

    'ORTHO-002': (
        'appointment scheduling rescheduling availability '
        'change appointment cancel appointment date time booking'
    ),

    'ORTHO-003': (
        'initial consultation first visit forms treatment goals '
        'evaluation consultation provider'
    ),

    'ORTHO-004': (
        'insurance payment cost price pricing financial financing '
        'treatment cost exact cost coverage reimbursement benefits'
    ),

    'ORTHO-005': (
        'patient portal login access password security code '
        'account identity verification authentication'
    ),

    'ORTHO-006': (
        'pain swelling broken wire broken appliance braces '
        'medication repair clinical emergency treatment advice'
    )
}


# Add metadata keywords to each chunk
chunks_df['metadata_keywords'] = chunks_df['doc_id'].map(
    document_keywords
)


# Improved search representation
chunks_df['search_text_v2'] = (
    chunks_df['title']
    + " "
    + chunks_df['title']
    + " "
    + chunks_df['metadata_keywords']
    + " "
    + chunks_df['metadata_keywords']
    + " "
    + chunks_df['chunk_text']
)


# Rebuild vectorizer
vectorizer_v2 = TfidfVectorizer(
    stop_words='english',
    ngram_range=(1, 2)
)

chunk_matrix_v2 = vectorizer_v2.fit_transform(
    chunks_df['search_text_v2']
)


def retrieve_chunks_v2(question, top_k=3):

    query_vector = vectorizer_v2.transform([question])

    scores = cosine_similarity(
        query_vector,
        chunk_matrix_v2
    ).flatten()

    top_indices = scores.argsort()[::-1][:top_k]

    results = chunks_df.iloc[top_indices].copy()

    results['score'] = scores[top_indices]

    return results[
        [
            'chunk_id',
            'doc_id',
            'title',
            'chunk_text',
            'score'
        ]
    ]


# ------------------------------------------------------------
# Re-test all 10 questions
# ------------------------------------------------------------

improved_rows = []

for _, row in test_cases.iterrows():

    retrieved = retrieve_chunks_v2(
        row['question'],
        top_k=3
    )

    top1 = retrieved.iloc[0]
    top2 = retrieved.iloc[1]

    improved_rows.append({
        'test_id': row['test_id'],
        'question': row['question'],
        'expected_source': row['expected_source'],
        'top1_source': top1['doc_id'],
        'top1_score': round(float(top1['score']), 4),
        'top2_source': top2['doc_id'],
        'top2_score': round(float(top2['score']), 4),
        'score_margin': round(
            float(top1['score'] - top2['score']),
            4
        )
    })


improved_retrieval_diagnostics = pd.DataFrame(
    improved_rows
)


# Save improvement evidence
improved_retrieval_diagnostics.to_csv(
    project_path
    / 'results'
    / 'improved_retrieval_diagnostics.csv',
    index=False
)


print("Improved retrieval diagnostics:")

improved_retrieval_diagnostics
```

```python
# ============================================================
# FINAL PROJECT STEP 2D
# Evidence Sufficiency + Hallucination Guard
# ============================================================

def check_evidence_sufficiency(question, retrieved_df):
    """
    Decide whether the retrieved evidence actually contains enough
    information to answer the user's question safely.
    """

    question_lower = question.lower()

    context = " ".join(
        retrieved_df['chunk_text'].tolist()
    ).lower()

    top_score = float(
        retrieved_df.iloc[0]['score']
    )

    # --------------------------------------------------------
    # Rule 1: Cancellation-fee questions
    # --------------------------------------------------------
    if "cancellation fee" in question_lower or (
        "fee" in question_lower
        and "appointment" in question_lower
    ):

        if (
            "cancellation fee" not in context
            and "fee" not in context
        ):
            return False, (
                "No cancellation-fee information "
                "appears in the retrieved sources."
            )

    # --------------------------------------------------------
    # Rule 2: Saturday availability
    # --------------------------------------------------------
    if "saturday" in question_lower:

        if "saturday" not in context:
            return False, (
                "Saturday availability is not documented "
                "in the retrieved sources."
            )

    # --------------------------------------------------------
    # Rule 3: Exact treatment-cost questions
    # --------------------------------------------------------
    if (
        "exactly how much" in question_lower
        or "treatment cost" in question_lower
        or "cost" in question_lower
    ):

        financial_evidence = [
            "exact treatment costs",
            "cost",
            "financial",
            "payment"
        ]

        if not any(
            term in context
            for term in financial_evidence
        ):
            return False, (
                "The retrieved sources do not contain "
                "sufficient financial information."
            )

    # --------------------------------------------------------
    # Rule 4: Very weak retrieval
    # --------------------------------------------------------
    if top_score < 0.12:
        return False, (
            "Retrieval confidence is too low."
        )

    return True, (
        "Retrieved evidence appears sufficient "
        "for a grounded response."
    )


# ============================================================
# Run Evidence Guard on All 10 Questions
# ============================================================

guard_rows = []

for _, row in test_cases.iterrows():

    retrieved = retrieve_chunks_v2(
        row['question'],
        top_k=3
    )

    sufficient, reason = check_evidence_sufficiency(
        row['question'],
        retrieved
    )

    # T7 and T10 are intentionally missing-information cases
    expected_sufficient = (
        row['test_id'] not in ['T7', 'T10']
    )

    guard_rows.append({
        'test_id': row['test_id'],
        'question': row['question'],
        'top1_source': retrieved.iloc[0]['doc_id'],
        'top1_score': round(
            float(retrieved.iloc[0]['score']),
            4
        ),
        'evidence_sufficient': sufficient,
        'expected_sufficient': expected_sufficient,
        'guard_correct': int(
            sufficient == expected_sufficient
        ),
        'guard_reason': reason
    })


evidence_guard_results = pd.DataFrame(
    guard_rows
)


# Save results
evidence_guard_results.to_csv(
    project_path
    / 'results'
    / 'evidence_guard_results.csv',
    index=False
)


print("Evidence Sufficiency Results:")

display(evidence_guard_results)


guard_accuracy = (
    evidence_guard_results[
        'guard_correct'
    ].mean() * 100
)

print(
    "\nEvidence Guard Accuracy:",
    round(guard_accuracy, 1),
    "%"
)
```

```python
# ============================================================
# FINAL PROJECT STEP 2E
# Safe RAG Response Logic
# ============================================================

def generate_safe_rag_response(question):
    """
    Retrieve evidence, check whether it is sufficient,
    then produce either a grounded response or a
    missing-information response.
    """

    # Retrieve top 3 chunks
    retrieved = retrieve_chunks_v2(
        question,
        top_k=3
    )

    # Check whether evidence is sufficient
    sufficient, reason = check_evidence_sufficiency(
        question,
        retrieved
    )

    top = retrieved.iloc[0]

    # --------------------------------------------------------
    # Missing-information response
    # --------------------------------------------------------
    if not sufficient:

        answer = (
            "Not enough information in the retrieved sources. "
            "Please contact the orthodontic office for confirmation."
        )

        return {
            'answer': answer,
            'answer_type': 'missing_information',
            'source': 'NONE',
            'source_chunk': 'NONE',
            'guard_reason': reason
        }

    # --------------------------------------------------------
    # Grounded response
    # --------------------------------------------------------
    answer = (
        f"Based on the retrieved information: "
        f"{top['chunk_text']} "
        f"Source: [{top['doc_id']} / {top['chunk_id']}]"
    )

    return {
        'answer': answer,
        'answer_type': 'grounded',
        'source': top['doc_id'],
        'source_chunk': top['chunk_id'],
        'guard_reason': reason
    }


# ============================================================
# Run all 10 final test cases
# ============================================================

final_response_rows = []

for _, row in test_cases.iterrows():

    result = generate_safe_rag_response(
        row['question']
    )

    final_response_rows.append({
        'test_id': row['test_id'],
        'question': row['question'],
        'expected_source': row['expected_source'],
        'expected_behavior': row['expected_behavior'],
        'answer_type': result['answer_type'],
        'returned_source': result['source'],
        'generated_answer': result['answer'],
        'guard_reason': result['guard_reason']
    })


final_rag_outputs = pd.DataFrame(
    final_response_rows
)


# Save final outputs
final_rag_outputs.to_csv(
    project_path
    / 'results'
    / 'final_rag_outputs.csv',
    index=False
)


print("Final safe RAG responses created:",
      len(final_rag_outputs))

display(final_rag_outputs)
```

```python
# ============================================================
# FINAL PROJECT STEP 3A
# Final 10-Case Evaluation
# ============================================================

evaluation_rows = []

missing_information_cases = ['T7', 'T10']

for _, row in test_cases.iterrows():

    test_id = row['test_id']

    final_row = final_rag_outputs[
        final_rag_outputs['test_id'] == test_id
    ].iloc[0]

    retrieval_row = improved_retrieval_diagnostics[
        improved_retrieval_diagnostics['test_id'] == test_id
    ].iloc[0]

    # --------------------------------------------------------
    # Metric 1: Retrieval correctness
    # T7 intentionally has no valid source, so retrieval is N/A.
    # --------------------------------------------------------

    if row['expected_source'] == 'NONE':
        retrieval_correct = None

    else:
        retrieval_correct = int(
            retrieval_row['top1_source']
            == row['expected_source']
        )

    # --------------------------------------------------------
    # Metric 2: Safe behavior correctness
    # --------------------------------------------------------

    expected_answer_type = (
        'missing_information'
        if test_id in missing_information_cases
        else 'grounded'
    )

    behavior_correct = int(
        final_row['answer_type']
        == expected_answer_type
    )

    # --------------------------------------------------------
    # Metric 3: Citation / source support
    # Only grounded answers should cite a source.
    # --------------------------------------------------------

    if expected_answer_type == 'grounded':

        citation_correct = int(
            row['expected_source']
            in final_row['generated_answer']
        )

    else:

        citation_correct = int(
            final_row['returned_source'] == 'NONE'
        )

    evaluation_rows.append({
        'test_id': test_id,
        'question': row['question'],
        'expected_source': row['expected_source'],
        'retrieved_source': retrieval_row['top1_source'],
        'answer_type': final_row['answer_type'],
        'retrieval_correct': retrieval_correct,
        'behavior_correct': behavior_correct,
        'citation_or_source_support_correct': citation_correct
    })


final_evaluation = pd.DataFrame(
    evaluation_rows
)


# Save detailed evaluation
final_evaluation.to_csv(
    project_path
    / 'results'
    / 'final_evaluation_scores.csv',
    index=False
)


# ============================================================
# Summary Metrics
# ============================================================

retrieval_valid = final_evaluation[
    final_evaluation['retrieval_correct'].notna()
]

retrieval_accuracy = (
    retrieval_valid['retrieval_correct'].mean()
    * 100
)

safe_behavior_accuracy = (
    final_evaluation['behavior_correct'].mean()
    * 100
)

citation_accuracy = (
    final_evaluation[
        'citation_or_source_support_correct'
    ].mean()
    * 100
)


final_summary_metrics = pd.DataFrame([{
    'number_of_tests': len(final_evaluation),
    'retrieval_accuracy_percent': round(
        retrieval_accuracy, 1
    ),
    'safe_behavior_accuracy_percent': round(
        safe_behavior_accuracy, 1
    ),
    'citation_source_support_percent': round(
        citation_accuracy, 1
    )
}])


final_summary_metrics.to_csv(
    project_path
    / 'results'
    / 'final_summary_metrics.csv',
    index=False
)


print("FINAL EVALUATION TABLE")
display(final_evaluation)

print("\nFINAL SUMMARY METRICS")
display(final_summary_metrics)
```

```python
# ============================================================
# FINAL PROJECT STEP 5A
# System Architecture Diagram
# ============================================================

import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(12, 7))

ax.set_xlim(0, 12)
ax.set_ylim(0, 10)
ax.axis('off')

def add_box(x, y, w, h, text):
    box = FancyBboxPatch(
        (x, y),
        w,
        h,
        boxstyle="round,pad=0.03",
        linewidth=1.5
    )
    ax.add_patch(box)

    ax.text(
        x + w/2,
        y + h/2,
        text,
        ha='center',
        va='center',
        fontsize=10
    )

def add_arrow(x1, y1, x2, y2):
    ax.annotate(
        '',
        xy=(x2, y2),
        xytext=(x1, y1),
        arrowprops=dict(
            arrowstyle='->',
            linewidth=1.5
        )
    )

# Main final-project pipeline
add_box(4.5, 8.5, 3, 0.8, "User Question")
add_box(4.5, 7.0, 3, 0.8, "TF-IDF + Domain Metadata")
add_box(4.5, 5.5, 3, 0.8, "Top-3 Retrieved Chunks")
add_box(4.5, 4.0, 3, 0.8, "Evidence Sufficiency Guard")
add_box(4.5, 2.5, 3, 0.8, "Grounded Answer or\nMissing-Information Response")
add_box(4.5, 1.0, 3, 0.8, "Source Citation + Evaluation")

add_arrow(6, 8.5, 6, 7.8)
add_arrow(6, 7.0, 6, 6.3)
add_arrow(6, 5.5, 6, 4.8)
add_arrow(6, 4.0, 6, 3.3)
add_arrow(6, 2.5, 6, 1.8)

# Knowledge base
add_box(
    0.5,
    5.5,
    2.5,
    1.2,
    "6 Fictional\nOrthodontic Documents"
)

add_arrow(
    3.0,
    6.1,
    4.5,
    6.1
)

# Local baseline comparison
add_box(
    8.8,
    5.4,
    2.7,
    1.4,
    "Baseline:\nLlama 3.2 3B\nvia Ollama"
)

add_box(
    8.8,
    3.2,
    2.7,
    1.2,
    "Baseline Evaluation\n10 Questions"
)

add_arrow(
    10.15,
    5.4,
    10.15,
    4.4
)

ax.text(
    6,
    9.7,
    "Final Orthodontic AI System Architecture",
    ha='center',
    va='center',
    fontsize=15,
    fontweight='bold'
)

ax.text(
    6,
    0.35,
    "Final system: retrieval-grounded assistant | Local Llama 3.2 3B used as baseline comparison",
    ha='center',
    fontsize=9
)

plt.tight_layout()

architecture_path = (
    project_path
    / 'images'
    / 'system_architecture.png'
)

plt.savefig(
    architecture_path,
    dpi=180,
    bbox_inches='tight'
)

plt.show()

print(
    "Architecture diagram saved to:",
    architecture_path
)
```
